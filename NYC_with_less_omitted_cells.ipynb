```python
# prompt: determine classes of unique values for each df column. Eliminate those classes from df with less than 5 frequency

from collections import defaultdict

# Create a dictionary to store the classes and their frequencies for each column
class_frequencies = defaultdict(lambda: defaultdict(int))

# Iterate through each column in the DataFrame
for column in df.columns:
  # Iterate through each unique value in the column
  for value in df[column].unique():
    # Count the frequency of the unique value
    class_frequencies[column][value] = df[column].value_counts().get(value, 0)

# Create a new DataFrame to store the filtered data
filtered_df = df.copy()

# Iterate through each column in the DataFrame
for column in df.columns:
  # Iterate through the unique values and their frequencies for the current column
  for value, frequency in class_frequencies[column].items():
    # If the frequency is less than 5, drop rows with this value
    if frequency <= 5:
      filtered_df = filtered_df[filtered_df[column] != value]

# Print the DataFrame with filtered classes
display(filtered_df.head())
```

```python
from sklearn.linear_model import Ridge
from sklearn.feature_selection import SequentialFeatureSelector
selected_features_forward_ridge = []
selected_features_backward_ridge = []

for k in range(1,len(X.columns)-1,1):
  print("For {} features SequentialFeatureSelector:".format(k))
  num_fs = k  # Desired number of features
  # Forward selection
  ridge = Ridge(alpha=0.1)
  sfs_forward = SequentialFeatureSelector(ridge, n_features_to_select=num_fs, direction='forward')
  sfs_forward.fit(X, y)
  selected_features_forward = X.columns[sfs_forward.get_support()]
  eliminated_features_forward = list(set(X.columns) - set(selected_features_forward))

  print("Forward Selection Ridge Regression:")
  print("Selected features:", list(selected_features_forward))
  print("Eliminated features:", eliminated_features_forward)

  selected_features_forward_ridge.append(selected_features_forward)

  # Backward selection
  sfs_backward = SequentialFeatureSelector(ridge, n_features_to_select=num_fs, direction='backward')
  sfs_backward.fit(X, y)
  selected_features_backward = X.columns[sfs_backward.get_support()]
  eliminated_features_backward = list(set(X.columns) - set(selected_features_backward))

  print("Backward Selection Ridge Regression:")
  print("Selected features:", list(selected_features_backward))
  print("Eliminated features:", eliminated_features_backward)

  selected_features_backward_ridge.append(selected_features_backward)

  print("\n")
```
